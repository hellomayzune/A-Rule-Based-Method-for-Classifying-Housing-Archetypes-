# Detached Houses: DT-R, DT-L and DT-V

```python
# Select "Detached" under Houses
dfd = df[(df.built_form == "Detached")]
```

```python
cond_DT = [
    dfd['total_floor_area'] <= 150, dfd['total_floor_area'] > 150
]
```

```python
choice_DT = ['DT-R', 'DT-L or DT-V']
```

```python
dfd['Archetype'] = np.select(cond_DT, choice_DT, default='Unknown')
```

```python
dfd.head(2)
```

```python
dfd.info()
```

```python
# get count per type
detached_archetype = dfd['Archetype'].value_counts()
print(detached_archetype)
```

# Semi-Detached Houses: SD-R, SD-S and SD-V

```python
# Select "Semi-Detached" under Houses
dfs = df[(df.built_form == "Semi-Detached")]
```

```python
cond_SD = [
    dfs['total_floor_area'] <= 100,
    (dfs['total_floor_area'] > 100) & (dfs['total_floor_area'] < 150),
    dfs['total_floor_area'] >= 150
]
```

```python
choice_SD = ['SD-R', 'SD-S', 'SD-V']
```

```python
dfs['Archetype'] = np.select(cond_SD, choice_SD, default='Unknown')
```

```python
dfs.tail(2)
```

```python
# get count per type
semi_archetype = dfs['Archetype'].value_counts()
print(semi_archetype)
```

# End-Terrace Houses: TR-E, TV-E and TG-E

```python
# Select "End-Terrace" under Houses
dfe = df[(df.built_form == "End-Terrace")]
```

```python
dfe['Archetype'] = np.select(cond_TE, choice_TE, default='Unknown')
```

## Limitations

This archetype classification relies on fixed rule-based thresholds rather than empirical or 
validated boundaries, and should be interpreted with the following caveats:

* **Fixed floor-area thresholds:** Cutoffs (100, 110, 150, and 200 m²) are defined heuristically rather than derived from statistical analysis of the underlying floor-area distribution. No histogram, quantile, or clustering analysis was performed, and the thresholds have not been validated against ground-truth archetype labels.

* **Hard boundaries:** Properties immediately above or below a threshold (e.g. 149 vs. 151 m² for detached houses) are assigned to different archetypes despite having almost identical characteristics. Boundary operators (`<=`, `<`, `>=`, `>`) are also applied inconsistently across dwelling types.

* **Unused storey-count information:** Although `flat_storey_count` was explored as a means of separating DT-L (1–2 storeys) from DT-V (3+ storeys), this logic was never incorporated into the final classification. Consequently, all large detached dwellings remain grouped as `DT-L or DT-V`.

* **Unused building attributes:** `construction_age_band` and `floor_height` are loaded into the dataset but are not used in the classification. As construction age strongly influences building fabric and energy performance, the resulting archetypes are based primarily on dwelling form and floor area rather than a broader building typology.

* **Limited property-type coverage:** The classification only handles houses and bungalows. Flats, maisonettes, park homes, and other property types are excluded from the final archetype dataset without reporting the number or proportion of omitted records.

* **No floor-area split for bungalows:** Bungalows are classified solely by `built_form` (detached versus non-detached), with no subdivision by floor area. This differs from the approach applied to detached, semi-detached, and terraced houses, where floor area is a primary classification variable.

* **Silent `Unknown` fallback:** Records that fail all classification conditions, including those with missing (`NaN`) `total_floor_area` values, default to `Unknown`. The workflow does not distinguish between missing data and genuinely unclassified properties or report their frequency.

* **No ground-truth validation:** The classification has not been evaluated against independently verified archetype labels. Assessment is limited to descriptive summaries (e.g. frequency counts and bar charts), so classification accuracy remains unknown.

* **Dataset-specific thresholds:** The classification rules were developed using a single South Yorkshire EPC dataset (April 2026 extract). Their applicability to other regions or future EPC datasets has not been assessed, and recalibration may be required for broader use.
