## Limitations

This archetype classification relies on fixed rule-based thresholds rather than empirical or 
validated boundaries, and should be interpreted with the following caveats:

- **Fixed gross_area thresholds**: Cutoffs (e.g. 100, 110, 150, 200 m²) are not derived from 
  statistical analysis of the underlying distribution and are not validated against ground-truth 
  archetype labels. No accuracy/confusion-matrix check has been performed.
- **Hard boundaries**: Properties just above/below a threshold (e.g. 149 vs 151 m²) are assigned 
  to different archetypes despite near-identical size; inequality operators (`<=`/`<`/`>=`) are 
  also applied inconsistently across categories.
- **premise_floor_count reliance for bungalow/house split**: Determines bungalow status alone; 
  bungalows are then classified by `premise_type` only, with no `gross_area` split — inconsistent 
  with the logic applied to other archetypes.
- **Unclassified records**: Properties with floor counts of 7–15, 18, or 32 are labelled only 
  `'Not_Bungalow'` and never receive a specific archetype (see `unknown_archetype` count above).
- **Manual floor-count corrections**: Anomalous values (`'3,5'`, `'2,8'`, `'2,3'`, `'1,3'`, `'1,2'`) 
  were recoded to single integers without documented justification, which may shift some 
  properties across the bungalow/house boundary.
- **Excluded outbuildings**: Rows tagged `"Domestic outbuilding"` are dropped, so the archetype 
  scheme covers only a subset of the building stock.
- **Single-variable proxy**: `gross_area` and `premise_type` cannot capture footprint shape, roof 
  form, bedroom count, build era, or extensions — all of which inform "true" architectural archetype.
- **No regional adjustment**: The same thresholds are applied across the dataset regardless of 
  regional variation in typical building size.
- **Static snapshot**: `gross_area` reflects a single point in time and doesn't account for later 
  extensions or conversions.
- **Silent `'Unknown'` fallback**: Records failing all conditions (e.g. missing `gross_area`) 
  default to `'Unknown'` without diagnostics on cause or frequency.
